### Visualizing the data

```python
from imutils import paths
import random
import matplotlib.pyplot as plt
import os
from PIL import Image
import cv2
import numpy as np
```

```python
# get all image paths from the folder in a list (local path)
imgPaths = list(paths.list_images(r"C:\Users\Arnava\.cache\kagglehub\datasets\pandianpandi\ppe-detection-dataset\versions\1\PPE_dataset\train\images"))

# function to display a grid of images
def display_images(image_paths, title, cols = 5, num_images = 25): # Set cols and add num_images
    random.shuffle(image_paths)
    num_images = min(len(image_paths), num_images)
    rows = (num_images + cols - 1) // cols

    plt.figure(figsize = (10, 10))
    plt.suptitle(title)

    for i, image_path in enumerate(image_paths[:num_images]):
        plt.subplot(rows, cols, i+1)
        img = Image.open(image_path)
        plt.imshow(img)
        plt.title(os.path.basename(os.path.dirname(image_path)))
        plt.axis("off")
    
    plt.tight_layout(rect = [0, 0.03, 1, 0.95])
    plt.show()

display_images(imgPaths, title = "Random Sample of training images.")
```
